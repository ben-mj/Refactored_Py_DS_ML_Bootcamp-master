### Create the virtual environment

```python
### 1. Create the virtual environment
# virtualenv --python=python<version ที่อยากใช้> venv_name
```

```python
### 2. Activate the virtualenv 
# source your-venv/bin/activate
```

```python
### 3. Add the virtualenv as a jupyter kernel
# ipython kernel install --user --name=your-venv
```

```python
### 4. List kernels >> to check ว่ามี kernels กี่ตัว
# Jupyter kernelspec list
```

```python
### 5. Remove kernel
# jupyter kernelspec remove <kernel-name>
```

```python
### 6. Quit virtual environment
# deactivate
```

```python
### 7. remove virtual environment
# rm -rf venv
```
